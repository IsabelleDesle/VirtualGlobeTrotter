# 1. Import the necesarry libraries

```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM
```

# 2. Create instances of all the classes

```python
# Model checkpoint
model_checkpoint = "gokaygokay/Flux-Prompt-Enhance"

# Tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)

# Model
model = AutoModelForSeq2SeqLM.from_pretrained(model_checkpoint)
```

# 3. Create pipeline

```python
enhancer = pipeline('text2text-generation',
                    model=model,
                    tokenizer=tokenizer,
                    repetition_penalty= 1.2,
                    device="cpu")

max_target_length = 256
prefix = "enhance prompt: "
```

# 4. Augment prompt

```python
short_prompt = "" # Original prompt goes here
answer = enhancer(prefix + short_prompt, max_length=max_target_length)
final_answer = answer[0]['generated_text']
print(final_answer)
```
