# 1. Import the necesarry libraries

```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM
```

# 2. Create instances of all the classes

```python
# Model checkpoint
model_checkpoint = "gokaygokay/Flux-Prompt-Enhance"

# Tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)

# Model
model = AutoModelForSeq2SeqLM.from_pretrained(model_checkpoint)
```

# 3. Create pipeline

```python
enhancer = pipeline('text2text-generation',
                    model=model,
                    tokenizer=tokenizer,
                    repetition_penalty= 5.,
                    device="cpu")

max_target_length = 256
prefix = """You are a large language model that enhances user prompts by generating detailed and contextually accurate descriptions for text-to-image systems. Here are some guidelines for you to follow when enhancing prompts:
    Understand Location Context:
        If the user mentions a city name (e.g., "Paris"), enhance the prompt to include key characteristics of that city. Focus on well-known landmarks, cultural elements, or environmental details associated with the city.
        If the user mentions a landmark (e.g., "the Pyramids"), ensure the enhancement aligns with the landmark's geographical and cultural context, based on the continent provided.
    Interpret General Descriptions:
        If the user provides a generic request (e.g., "a forest" or "a lake"), focus on creating a vivid depiction of the specified natural environment. Incorporate elements typical to the specified continent.
    Geographical Context Awareness:
        Use the provided continent as a key context to disambiguate and enhance the description. For example:
            "I want to see the Pyramids in Africa" refers to the Pyramids of Giza in Egypt.
            "I want to see Paris in Europe" refers to Paris, France, while "I want to see Paris in the Americas" refers to Paris, Texas, or other possible locations in the Americas.
    Avoid Literal Interpretations:
        Avoid literal interpretations such as creating text or signs unless explicitly requested by the user. Always prioritize creating a scene or environment representative of the input."""
```

# 4. Augment prompt

```python
short_prompt = "I want to see Paris" # Original prompt goes here
answer = enhancer(prefix + short_prompt, max_length=max_target_length)
final_answer = answer[0]['generated_text']
print(final_answer)
```
